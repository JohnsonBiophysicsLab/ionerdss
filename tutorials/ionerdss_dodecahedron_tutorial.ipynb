# ioNERDSS Tutorial: Synthetic Dodecahedron Workflow

This notebook builds a model without a PDB structure:

1. Build a synthetic dodecahedron with `build_system_from_plat(...)`.
2. Export `nerdss_files/` and then run `nerdss -f parms.inp`.


## Imports


```python
from pathlib import Path
import subprocess

from ionerdss import build_system_from_plat
```

## 1. Build a synthetic dodecahedron and run NERDSS

`build_system_from_plat(...)` returns the system and its reaction rules, and with `output_nerdss=True` also writes the NERDSS input files.


```python
save_folder = Path("tutorials_output/dode_tutorial")
save_folder.mkdir(parents=True, exist_ok=True)

system, rxn = build_system_from_plat(
    solid_type="dode",
    radius=8,
    sigma=1,
    output_nerdss=True,
    output_dir=str(save_folder),
)

print(f"Built {len(system.molecule_types.molecule_types)} molecule type(s)")
print(f"Built {len(rxn)} reaction rule(s)")
```

```python
import shutil

from ionerdss.nerdss_simulation import resolve_nerdss_executable

# The NERDSS executable, or your NERDSS checkout (bin/nerdss inside it is found). It is
# picked up from your PATH if `nerdss` is on it; otherwise replace the placeholder.
nerdss_exe = resolve_nerdss_executable(shutil.which("nerdss") or "PATH_TO_NERDSS_REPO")

print("\nNow running NERDSS simulation...\n")

nerdss_dir = save_folder / "nerdss_files"

result = subprocess.run(
    [str(nerdss_exe), "-f", "parms.inp"],
    cwd=nerdss_dir,
    capture_output=True,
    text=True,
    check=True,
)
# NERDSS reports on every step; show only how the run ended.
print("\n".join(result.stdout.splitlines()[-4:]))
```

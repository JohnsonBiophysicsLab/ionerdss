# ioNERDSS Tutorial: 6BNO and Synthetic Dodecahedron Workflows

This notebook shows two end-to-end setup patterns:


1. Build a synthetic dodecahedron with `build_system_from_plat(...)`.
2. export `nerdss_files/` and then run `nerdss -f parms.inp`.


## Imports

The first workflow uses the explicit PDB builder API so you can see where the hyperparameters are applied.


```python
from pathlib import Path
import subprocess

from ionerdss import build_system_from_pdb, build_system_from_plat
from ionerdss.model.pdb.main import PDBModelBuilder
from ionerdss.model.pdb.hyperparameters import PDBModelHyperparameters
from ionerdss.model.pdb.parser import PDBParser
```

## 1. Build a synthetic dodecahedron and run NERDSS

This matches the requested `build_system_from_plat(...)` workflow.


```python
save_folder = Path("tutorials_output/dode_tutorial")
save_folder.mkdir(parents=True, exist_ok=True)

sys, rxn = build_system_from_plat(
    solid_type="dode",
    radius=8,
    sigma=1,
    output_nerdss=True,
    output_dir=str(save_folder),
)

print(f"Built {len(sys.molecule_types.molecule_types)} molecule type(s)")
print(f"Built {len(rxn)} reaction rule(s)")
```

```python
# Replace with your local nerdss path 
nerdss_cmd = "~/Workspace/nerdss_development/bin/nerdss -f parms.inp"

print("\nNow running NERDSS simulation...\n")

nerdss_dir = save_folder / "nerdss_files"

subprocess.run(
    nerdss_cmd,
    shell=True,
    cwd=nerdss_dir,
    executable="/bin/bash",
    check=True,
)
```

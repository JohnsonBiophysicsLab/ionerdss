# Quick Start - Minimal Example

This is the simplest way to use ionerdss. Just one function call generates NERDSS simulation files from a PDB ID.

```python
# minimalistic example
from ionerdss import build_system_from_pdb

pdb_id = "6bno"

# Simplified API - single function call
system = build_system_from_pdb(
    source=pdb_id,
    workspace_path=f"{pdb_id}_dir",
)
```

## What This Does

The `build_system_from_pdb()` function:
- Downloads the PDB structure
- Detects interfaces between chains
- Generates coarse-grained molecular models
- Exports NERDSS simulation files to `6bno_dir/nerdss_files/`

For more control over the process, see the full tutorial `ionerdss_tutorial_6bno.ipynb`.

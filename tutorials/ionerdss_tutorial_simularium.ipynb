# ioNERDSS Simularium Visualization Tutorial

This tutorial demonstrates how to convert NERDSS trajectory outputs into a format that can be viewed in [Simularium](https://simularium.allencell.org/), a free 3D viewer for spatial biological simulations.

## Installation Requirement 
To use this visualization tool, you **must install `ioNERDSS` with the `simularium` optional dependency** because `simulariumio` (and its large sub-dependencies) is omitted from the base installation to keep things lightweight.

Run the following command in your terminal or Jupyter environment's command prompt:
```bash
pip install "ioNERDSS[simularium]"
```
*(If you are compiling from source code, you can run `pip install -e ".[simularium]".)*

### Known compatibility issue: `simulariumio` on PyPI vs. GitHub

> The newest `simulariumio` release on PyPI is **1.13.0** (2025-08-18). Its NERDSS converter writes in place into an array that **pandas 3.0+ returns read-only**, so conversion fails with:
> ```
> ValueError: assignment destination is read-only
> ```
> The fix exists on the [simulariumio GitHub `main` branch](https://github.com/simularium/simulariumio) (commit `ecfd75b`, 2026-04-01) but has not been released to PyPI yet.
>
> The `simularium` extra therefore constrains `pandas<3`, which avoids the problem on a clean install. If you already have pandas 3.x in your environment, either pin pandas:
> ```bash
> pip install "pandas<3"
> ```
> or install `simulariumio` straight from GitHub:
> ```bash
> pip install "git+https://github.com/simularium/simulariumio.git@main"
> ```

````python
# Import the library
import ionerdss
import logging

# We set up logging so we can see warnings if simulariumio isn't properly installed
logging.basicConfig(level=logging.INFO)
````

## Step 1: Simulate your System (or use an existing workspace)

This tutorial ships with a ready-made NERDSS output directory so it runs as-is: `../data/8y7s_dir/nerdss_output/1`, relative to this notebook.

BELOW: you must provide the path to a simulation output directory in the variable `simulation_output_dir`.

This directory must minimally contain the nerdss output files of: 
- `.mol` structural files
- `parms.inp` (the parameter configurations)
- `PDB/` directory with `.pdb` frame structures inside.

*To visualize your own run instead, point `simulation_output_dir` at its output folder — for example `6bno_dir/nerdss_files` after running `ionerdss_tutorial_6bno.ipynb`, which explains how to create one.*

BELOW: you must also provide an output name for the simularium file in the variable `output_filename`. For example:
output_filename = "8y7s_trajectory"

The function will append the .simularium filetype to the end of this output_filename.

````python
# You must define where your NERDSS simulation output lives!
# This is the example run bundled with the repository, relative to this notebook.
# Replace it with a path to your own NERDSS output folder (absolute paths work too).
simulation_output_dir = "../data/8y7s_dir/nerdss_output/1"

# Define the name of the file you want to output (do not include the extension here)
output_filename = "8y7s_trajectory"
````

## Step 2: Convert to Simularium Format

Next, we simply use the `convert_simularium` function! It will read the `.mol` files for structure information, the `parms.inp` file for the box dimensions and timestamps, and the `PDB/*.pdb` frames for the coordinates.


````python
try:
    ionerdss.convert_simularium(
        input_dir=simulation_output_dir,
        output_name=output_filename,
        output_format='binary'  # You can also choose 'json', but binary makes smaller files.
    )
    print(f"Success!")
except Exception as e:
    print(f"Error occurred during conversion: {e}")
````

## Step 3: Visualize

1. Go to [simularium.allencell.org/viewer](https://simularium.allencell.org/viewer/).
2. Drag and drop the `.simularium` file into the viewer or click the `load model` button to upload via file finder.

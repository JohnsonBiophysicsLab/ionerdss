# ioNERDSS Trajectory Movie Tutorial

This tutorial turns a finished NERDSS run into a GIF or MP4 movie with `ionerdss.render_trajectory_movie`.

Every molecule is drawn as a shaded sphere at its center of mass:

- the camera and the simulation box (`WaterBox` in `parms.inp`) stay fixed for the whole movie;
- the time (`iteration × timeStep`) is printed large in one corner, in one unit for the whole movie;
- each molecule type has its own color, whether the molecule is free or in a complex;
- each sphere's radius is the one ionerdss used to compute the molecule's `D` and `Dr`.

No extra dependencies are needed: `pip install ioNERDSS` is enough. Writing `.mp4` also needs [ffmpeg](https://ffmpeg.org/) (`conda install ffmpeg` or `pip install imageio-ffmpeg`).

```python
import logging

import ionerdss

# Show which radius each molecule type gets, and any warnings.
logging.basicConfig(level=logging.INFO)
```

## Step 1: Point at a NERDSS run directory

A run directory is the folder NERDSS ran in. It holds `parms.inp`, one `.mol` file per molecule type, and the saved frames: `PDB/<iteration>.pdb` and/or `DATA/trajectory.xyz`. If you ran the 6bno tutorial, that is `6bno_dir/nerdss_files`.

```python
run_dir = "./6bno_dir/nerdss_files"
```

## Step 2: Render the movie

The output format follows the file extension. Useful options:

- `frame_stride=n` renders every n-th saved frame, for long runs;
- `show_legend=True` lists each molecule type's color and radius next to the time;
- `time_font_size=` sets the height of the time label in pixels;
- `radii={"A": 2.0}` overrides a sphere radius in nm, and `view=(azimuth, elevation)` turns the camera;
- `n_jobs=-1` renders frames on every core, which can help for very large systems.

```python
movie = ionerdss.render_trajectory_movie(
    run_dir,
    "./6bno_dir/6bno_trajectory.gif",  # or "./6bno_dir/6bno_trajectory.mp4" (needs ffmpeg)
    fps=5,
)
print(f"Saved {movie}")
```
